#### 1. Copy on write
#### 2. Merge on read

##### Copy on write

```sql
CREATE TABLE deltacatalog.deltadb.invoices_copyonw
TBLPROPERTIES ('delta.enableDeletionVectors' = false)
AS
select * from parquet.`abfss://labdata@stgdatabricksdelta.dfs.core.windows.net/invoices/invoices_1_100.parquet`
```

```sql
update deltacatalog.deltadb.invoices_copyonw
set quantity = 100
where customer_id = 10
```

```sql
delete from deltacatalog.deltadb.invoices_copyonw
where customer_id = 25
```

```sql
desc history deltacatalog.deltadb.invoices_copyonw
```

##### Merge on read

```sql
CREATE TABLE deltacatalog.deltadb.invoices_mergeor
TBLPROPERTIES ('delta.enableDeletionVectors' = true)
AS
select * from parquet.`abfss://labdata@stgdatabricksdelta.dfs.core.windows.net/invoices/invoices_1_100.parquet`
```

```sql
desc extended deltacatalog.deltadb.invoices_mergeor
```

```sql
delete from deltacatalog.deltadb.invoices_mergeor
where customer_id = 25
```

```sql
update deltacatalog.deltadb.invoices_mergeor
set quantity = 100
where customer_id = 10
```

```sql
optimize deltacatalog.deltadb.invoices_mergeor;
```

```sql
set spark.databricks.delta.retentionDurationCheck.enabled = false;
vacuum deltacatalog.deltadb.invoices_mergeor retain 0 hours;
```
